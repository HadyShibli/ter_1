```python
import json
import ipywidgets as w
from IPython.display import display

PATH = '/content/drive/My Drive/students.json'

# Deserialization – read the students from the file
with open(PATH, 'r', encoding='utf-8') as fid:
    students = json.load(fid)

# ---------- design ----------
display(w.HTML('''
<style>
  .card {background: linear-gradient(135deg,#fff8e1,#ffe0b2); border-radius:14px;
         padding:18px 22px; box-shadow:0 4px 12px rgba(0,0,0,.15); font-family:Arial;}
  .card h2 {margin:0 0 6px 0; color:#e65100;}
  .card p  {margin:0; color:#555;}
</style>
<div class="card"><h2>🎓 Students Form</h2><p>Choose a student to see the details and update the favorite show</p></div>
'''))

style = {'description_width': '130px'}
lay = w.Layout(width='480px')

pick    = w.Dropdown(options=[(f"{s['first_name']} {s['last_name']}", i) for i, s in enumerate(students)],
                     description='Student:', style=style, layout=lay)
first   = w.Text(description='First name:', disabled=True, style=style, layout=lay)
last    = w.Text(description='Last name:',  disabled=True, style=style, layout=lay)
email   = w.Text(description='Braude email:', disabled=True, style=style, layout=lay)
courses = w.Textarea(description='Courses:', disabled=True, style=style, layout=w.Layout(width='480px', height='90px'))
link    = w.HTML()
fav     = w.Text(description='⭐ Favorite show:', placeholder='Type the favorite show...', style=style, layout=lay)
btn     = w.Button(description='Update', button_style='warning', icon='save', layout=w.Layout(width='150px'))
msg     = w.HTML()

def show_student(change=None):
    s = students[pick.value]
    first.value = s['first_name']
    last.value  = s['last_name']
    email.value = s['email']
    courses.value = '\n'.join(s['courses'])
    link.value = f"<b>Interesting link:</b> <a href='{s['link']}' target='_blank'>{s['link']}</a>"
    fav.value = s.get('favorite_show', '')
    msg.value = ''

def update_student(b):
    if not fav.value.strip():
        msg.value = "<span style='color:red'>✖ Please type a favorite show first</span>"
        return
    students[pick.value].update({"favorite_show": fav.value.strip()})
    with open(PATH, 'w', encoding='utf-8') as fid:
        json.dump(students, fid, ensure_ascii=False, indent=4)
    msg.value = f"<span style='color:green'>✔ Saved for {students[pick.value]['first_name']} in students.json</span>"

pick.observe(show_student, names='value')
btn.on_click(update_student)
show_student()

display(w.VBox([pick, first, last, email, courses, link, fav, btn, msg],
               layout=w.Layout(border='2px solid #ffb74d', padding='14px', border_radius='12px', width='540px')))
```
